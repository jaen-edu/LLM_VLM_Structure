### 학습 흐름 요약

```
[텍스트의 흐름]
문자열
  →  Tokenizer (BPE 서브워드 분할)
  →  Token IDs
  →  Embedding (정수 → d_model 차원 벡터)
  →  Positional Encoding (위치 정보 주입)
  →  [N × Transformer Block]
       ├─ Multi-Head Self-Attention  ← Q·K·V 유사도로 토큰 간 관계 포착
       ├─ Feed-Forward Network
       └─ Layer Norm + Residual
  →  LM Head(Language Modeling Head) (d_model → vocab_size)
  →  Softmax → 확률 분포
  →  디코딩 전략 (Greedy / Temperature / Top-p)
  →  다음 토큰 선택 → 반복 (Autoregressive)

[이미지의 흐름]
이미지(H×W×3)
  →  패치 분할 (P×P 크기)
  →  Linear Projection (각 패치 → d_model 벡터)
  →  [CLS] + Positional Encoding
  →  Transformer Encoder
  →  이미지 임베딩 시퀀스
```

이 요약에서 중요한 것은, 텍스트와 이미지가 완전히 다른 것처럼 보여도 둘 다 결국 **토큰 시퀀스 또는 임베딩 시퀀스**로 바뀐다는 점이다.  
텍스트는 BPE와 임베딩을 거쳐 Transformer로 들어가고, 이미지는 패치 분할과 projection을 거쳐 Transformer로 들어간다. 즉, 1일차의 핵심은 "입력이 달라도 Transformer가 다룰 수 있는 표현으로 바꾸는 과정"을 이해하는 데 있다.

### 혼자 점검하기 모범 답안

**1. 1일차의 텍스트 흐름과 이미지 흐름은 각각 어떻게 요약할 수 있는가?**

텍스트 흐름은 문자열 → 토큰화 → 임베딩 → Transformer → LM Head(Language Modeling Head) → 다음 토큰 선택으로 요약할 수 있다. 이미지 흐름은 이미지 → 패치 분할 → 패치 임베딩 → Transformer Encoder → 이미지 임베딩 시퀀스로 요약할 수 있다.

**2. `Patch Token`과 `BPE Tokenizer`는 어떤 점에서 비슷하고 어떤 점에서 다른가?**

둘 다 원본 입력을 모델이 다룰 수 있는 작은 단위로 나눈다는 점에서는 비슷하다. 다만 BPE는 텍스트를 서브워드 단위로 나누고, Patch Token은 이미지를 고정 크기 패치 단위로 나눈다는 점이 다르다.
